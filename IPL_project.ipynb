***Q10. Convert the date column to the appropriate datetime format.***

```python
# Replacing the inconsistent date with the correct format
df['date'] = df['date'].str.replace('April11,2022', 'April 11,2022')

#Changing data type
df['date'] = pd.to_datetime(df['date'], format='%B %d,%Y')
```

### Now, Here comes some Basic Questions

* Which team won the most matches?

```python
df.groupby('match_winner')['match_winner'].count().sort_values(ascending = False).head(1)

#Gujarat won the most matches
```

* Toss Decision Trends

```python
df.groupby([df['date'].dt.year, 'toss_decision'])['toss_decision'].count()

#In this particular data set, in the year 2022, the toss decision shows a shift towards choosing fielding after winning the toss.
```

* Toss Winner vs Match Winner

```python
toss = df[['team1', 'team2', 'toss_winner', 'match_winner']]
toss['toss_match_same'] = df['toss_winner'] == df['match_winner']
toss.groupby('toss_match_same')['toss_match_same'].count()

#That means the toss winner also won 36 matches, while in 80 matches the toss winner lost.
```

* How do teams win? (Runs vs Wickets)

```python
df.groupby(['match_winner', 'won_by'])['won_by'].count()

# Bangalore won 5 matches by runs and 4 matches by wickets, and similarly for other teams.
```

# Visualization Section 📊

***1). Top 10 individual scorers***

```python
top_scorers = df.sort_values(by = 'highscore', ascending = False).head(10)

sns.barplot(data = top_scorers, x='top_scorer', y='highscore', color='orange')

plt.xticks(rotation=45)
plt.title('Top 10 Individual Scores')
plt.xlabel('Players')
plt.ylabel('Runs')
```

***2). Top 10 venues by number of matches***

```python
top_venues = df.groupby('venue')['match_id'].count().sort_values(ascending = False).head(10)

sns.barplot(x = top_venues.index, y = top_venues.values, color='purple')

plt.xticks(rotation = 45)
plt.title('Top 10 Venues by Number of Matches')
plt.xlabel('Venue')
plt.ylabel('Number of Matches')
```

***3). Matches won by runs vs wickets***

```python
sns.countplot(data = df, x = df['won_by'], color = 'cyan')

plt.title('Matches Won by Runs vs Wickets')
plt.xlabel('Won By')
plt.ylabel('Number of Matches')
plt.show()
```

***4). Distribution of winning margins by Runs***

```python
runs_wins = df[df['won_by'] == 'Runs']

sns.histplot(data=runs_wins, x='margin', bins = 5, color = 'red')

plt.title('Distribution of Winning Margins by Runs')
plt.xlabel('Winning Margin (Runs)')
plt.ylabel('Number of Matches')
```

***5). First innings vs second innings score***

```python
sns.scatterplot(
    data=df,
    x='first_ings_score',
    y='second_ings_score',
    color = 'green'
)

plt.title('First Innings Score vs Second Innings Score')
plt.xlabel('First Innings Score')
plt.ylabel('Second Innings Score')
```
