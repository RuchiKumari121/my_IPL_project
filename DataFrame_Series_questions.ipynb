```python
#Online Store

import numpy as np
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Name": ["Ravi", "Priya", "Amit", "Neha", "Karan"],
    "City": ["Delhi", "Mumbai", "Delhi", None, "Pune"]
})

orders = pd.DataFrame({
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006],
    "Customer_ID": [101, 102, 101, 103, 105, 102],
    "Amount": [2500, 1800, 3200, np.nan, 4500, 2200]
})

print(customers.isnull())
print(customers.isnull().sum())
customers["City"] = customers["City"].fillna("Unknown")

print(orders.isnull())
print(orders.isnull().sum())
orders["Amount"] = orders["Amount"].fillna(orders["Amount"].mean())

pd.merge(customers, orders, on = "Customer_ID", how = "inner")

orders.groupby('Customer_ID')["Amount"].sum()


#Series wala question
customer_total = (
    pd.merge(customers, orders, on="Customer_ID", how="inner")
      .groupby("Name")["Amount"]
      .sum()
)
new_series = pd.Series(
    # orders.groupby('Customer_ID')["Amount"].sum(),
    customer_total.values,
    index = customer_total.index.get_level_values("Name") 
)

new_series
```

```python
#Movie Streaming platform

january = pd.DataFrame({
    "Movie_ID": [1, 2, 3, 4],
    "Movie": ["Avatar", "Inception", "Joker", "Titanic"],
    "Views": [1200, 1500, np.nan, 900],
    "Rating": [4.5, 4.8, 4.2, np.nan]
})

february = pd.DataFrame({
    "Movie_ID": [5, 6, 7, 8],
    "Movie": ["Interstellar", "Dune", "Avengers", "Frozen"],
    "Views": [1800, 1400, np.nan, 1100],
    "Rating": [4.9, 4.6, 4.4, 4.1]
})

print(january.isnull())
print(february.isnull())

january["Views"] = january["Views"].fillna(0)
february["Views"] = february["Views"].fillna(0)

print(january)
print(february)

january["Rating"] = january["Rating"].fillna(january["Rating"].mean())

january["Month"] = "January"
february["Month"] = "February"

print(january)
print(february)


pd.concat([january, february]).reset_index()
```

```python
#Hospital

patients = pd.DataFrame({
    "Patient_ID": [1, 2, 3, 4, 5, 6],
    "Name": ["Rahul", "Anita", "Suresh", "Pooja", "Vikas", "Meena"],
    "Doctor_ID": [101, 102, 101, 103, 104, 102],
    "Age": [25, 34, np.nan, 45, 29, np.nan]
})

doctors = pd.DataFrame({
    "Doctor_ID": [101, 102, 103, 104],
    "Doctor": ["Dr. Sharma", "Dr. Mehta", "Dr. Khan", "Dr. Singh"],
    "Department": ["Cardiology", "Neurology", "Orthopedic", "Dermatology"]
})

followup = pd.DataFrame({
    "Patient_ID": [7, 8],
    "Name": ["Arjun", "Kavita"],
    "Doctor_ID": [103, 104],
    "Age": [38, 31]
})

print(patients.isnull())

patients["Age"] = patients["Age"].fillna(patients["Age"].mean())

patient_followup = pd.concat([patients, followup])
patient_followup.reset_index(drop = True, inplace = True)


merged_data = pd.merge(patient_followup, doctors, on = "Doctor_ID", how = "inner")

print(merged_data.groupby('Doctor')["Patient_ID"].count())

print(merged_data.groupby('Department')["Patient_ID"].count())
```
